```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
```

```python
df = pd.DataFrame(data=iris.data,columns=iris.feature_names)
```

```python
df
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)
```

```python
scaled_data
```

```python
scaled_new = scaled_data[:,0:2]
```

```python
plt.scatter(scaled_new[:,0],scaled_new[:,1])
```

```python
from sklearn.cluster import DBSCAN
dbscan = DBSCAN(eps=0.45)
dbscan.fit(scaled_new)
```

```python
plt.scatter(scaled_new[:,0],scaled_new[:,1],c=dbscan.labels_)
```

```python
from sklearn.metrics import silhouette_score
score = silhouette_score(X=scaled_new,labels=dbscan.labels_)
```
