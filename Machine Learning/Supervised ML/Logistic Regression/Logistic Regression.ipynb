### Logistic Rregression

```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
from sklearn.datasets import make_classification
```

```python
X,y = make_classification(n_samples=1000,n_features=10,n_classes=2,random_state=42)
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test, y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
logistic = LogisticRegression()
```

```python
logistic = logistic.fit(X_train,y_train)
```

```python
y_pred = logistic.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

```python
score = accuracy_score(y_test,y_pred)
print(score)
cm= confusion_matrix(y_test,y_pred)
print(cm)
print(classification_report(y_test,y_pred))
```

### Hyperparameter Tunning using GridSearchCV

```python
model = LogisticRegression()
penalty = ['l1','l2','elasticnet']
c_values = [100,10,1,0.0,0.1,0.01]
solver = ['newton-cg','lbfgs','liblinear','saga']
```

```python
params =dict(penalty=penalty,C=c_values,solver=solver)
```

```python
params
```

```python
#GridSearchCv
from sklearn.model_selection import GridSearchCV
```

```python
from sklearn.model_selection import StratifiedKFold
cv = StratifiedKFold()
```

```python
grid =GridSearchCV(estimator=model,param_grid=params,scoring='accuracy',cv=cv)
```

```python
grid
```

```python
grid.fit(X_train,y_train)
```

```python
y_pred = grid.predict(X_test)
```

```python
score = accuracy_score(y_test,y_pred)
print(score)
cm= confusion_matrix(y_test,y_pred)
print(cm)
print(classification_report(y_test,y_pred))
```

```python
grid.best_params_
```

### RandomizedSearchCV

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
model = LogisticRegression()
rcv = RandomizedSearchCV(estimator=model,param_distributions=params,scoring='accuracy',cv=cv)
```

```python
rc = rcv.fit(X_train,y_train)
```

```python
y_pred = rc.predict(X_test)
```

```python
score = accuracy_score(y_test,y_pred)
print(score)
cm= confusion_matrix(y_test,y_pred)
print(cm)
print(classification_report(y_test,y_pred))
```

```python
rc.best_params_
```

## Imbalanced Dataset

```python
X,y = make_classification(n_samples=10000,n_features=10,n_clusters_per_class=1,n_redundant=0,weights=[0.99],n_classes=2,random_state=42)
```

```python
from collections import Counter
```

```python
Counter(y)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test, y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
model = LogisticRegression()
penalty = ['l1','l2','elasticnet']
c_values = [100,10,1,0.0,0.1,0.01]
solver = ['newton-cg','lbfgs','liblinear','saga']
weights = [{0:w,1:y}for w in [1,10,50,100] for y in [1,10,50,100]]
```

```python
params = dict(penalty=penalty,C=c_values,solver=solver,class_weight=weights)
```

```python
grid  = GridSearchCV(estimator=model,param_grid=params,scoring='accuracy',cv=cv)
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
y_pred = grid.predict(X_test)
```

```python
score = accuracy_score(y_test,y_pred)
print(score)
cm= confusion_matrix(y_test,y_pred)
print(cm)
print(classification_report(y_test,y_pred))
```
