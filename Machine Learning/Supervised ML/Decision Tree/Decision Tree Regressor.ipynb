### Decision Tree Regression

```python
#importing important libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Loadind dataset

```python
from sklearn.datasets import load_diabetes
```

```python
diabetes = load_diabetes()
```

```python
diabetes
```

```python
X = pd.DataFrame(diabetes['data'],columns=diabetes.feature_names)
```

```python
X
```

```python
Y = pd.DataFrame(diabetes['target'])
```

```python
Y
```

```python
#train test split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,Y,test_size=0.20,random_state=10)
```

### Model Implementation

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
dtr = DecisionTreeRegressor()
```

```python
dtr.fit(X_train,y_train)
```

```python
y_pred = dtc.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
print('Score: ', r2_score(y_pred,y_test))
print('Score: ', mean_squared_error(y_pred,y_test))
print('Score: ', mean_absolute_error(y_pred,y_test))
```

### Hyperparameter Tunning

```python
param_grid = {
        'criterion': ['squared_error','friedman_mse','absolute_error'],
        'splitter':  ['best','random'],
        'max_depth': [1,2,3,4,5,10,15,20,25],
        'max_features': ['auto','sqrt','log2']
}
```

```python
from sklearn.model_selection import GridSearchCV
grid = GridSearchCV(DecisionTreeRegressor(),param_grid=param_grid,scoring='neg_mean_squared_error')
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
y_pred = grid.predict(X_test)
```

```python
print('Score: ', r2_score(y_pred,y_test))
print('Score: ', mean_squared_error(y_pred,y_test))
print('Score: ', mean_absolute_error(y_pred,y_test))
```
