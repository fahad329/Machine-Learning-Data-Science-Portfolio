```python
df['Reviews'].unique()
```

```python
df_copy = df.copy()
```

```python
df_copy = df_copy.drop(df_copy.index[10472])
```

```python
df_copy[~df_copy['Reviews'].str.isnumeric()]
```

```python
df_copy['Reviews'] = df_copy['Reviews'].astype('int')
```

```python
df_copy.info()
```

```python
df_copy['Size'].unique()
```

```python
def fix_size(s):
  try:
    if s.endswith('M'):
      return float(s[:-1])
    elif s.endswith('k'):
      return (float(s[:-1]) / 1024)
    elif s == '1,000+':
      return 1.0
    else:
      return np.nan
  except:
      return np.nan
     
```

```python
df_copy['Size']= df_copy['Size'].apply(fix_size)
```

```python
df_copy.head()
```

```python
df_copy['Installs'].unique()
```

```python
df_copy['Installs'] = df_copy['Installs'].str.replace('+','')
df_copy['Installs'] = df_copy['Installs'].str.replace(',','')
```

```python
df_copy['Installs'].unique()
```

```python
df_copy.head()
```

```python
df_copy['Price'].unique()
```

```python
df_copy['Price']= df_copy['Price'].str.replace('$','')
df_copy['Price']= df_copy['Price'].str.replace(',','')
```

```python
df_copy['Price'].unique()
```

```python
df_copy.tail()
```

```python
df_copy['Installs'] =df_copy['Installs'].astype(int)
df_copy['Price'] =df_copy['Price'].astype(float)
```

```python
df_copy.info()
```

```python
df_copy['Last Updated'].unique()
```

```python
df_copy['Last Updated'] = pd.to_datetime(df_copy['Last Updated'])
```

```python
df_copy['Day'] = df_copy['Last Updated'].dt.day
df_copy['Month'] = df_copy['Last Updated'].dt.month
df_copy['Year'] = df_copy['Last Updated'].dt.year
```

## 3.2 Feature Information
1. App :- Name of the App
2. Category :- Category under which the App falls.
3. Rating :- Application's rating on playstore
4. Reviews :- Number of reviews of the App.
5. Size :- Size of the App.
6. Install :- Number of Installs of the App
7. Type :- If the App is free/paid
8. Price :- Price of the app (0 if it is Free)
9. Content Rating :- Appropiate Target Audience of the App.
10. Genres:- Genre under which the App falls.
11. Last Updated :- Date when the App was last updated
12. Current Ver :- Current Version of the Application
13. Android Ver :- Minimum Android Version required to run the App

```python
## Proportion of count data on categorical columns

for col in categorical_features:
    print(df_copy[col].value_counts(normalize=True)*100)
```

```python
plt.figure(figsize=(15,15))
plt.suptitle('Univariate Analysis of Numerical Features', fontsize=20, fontweight='bold', alpha=0.8, y=1.)

for i in range(0,len(numerical_features)):
    plt.subplot(5,3,i+1)
    sns.kdeplot(x=df_copy[numerical_features[i]],shade=True, color='r')
    plt.xlabel(numerical_features[i])
    plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(20,15))
plt.suptitle('Univariate Analysis of Categorical Features', fontsize=20, fontweight='bold', alpha=0.8, y=1.)
categories = ['Type', 'Content Rating']
for i in range(0,len(categories)):
    plt.subplot(2,2,i+1)
    sns.countplot(x=df_copy[categories[i]],palette='Set2')
    plt.xlabel(categories[i])
    plt.tight_layout()
plt.show()
```

## 1. Which Category has largest number of installations??

```python
df_copy.head(2)
```

```python
cat_install = df_copy.groupby(['Category'])['Installs'].sum().sort_values(ascending=False).reset_index()
cat_install.Installs = cat_install.Installs/1000000000
```

```python
df_install = cat_install.head(10)
```

```python
df_install
```

```python
plt.figure(figsize=(15,8))
#sns.set_context('talk')
sns.set_style('darkgrid')
ax=sns.barplot(x=df_install['Installs'], y=df_install['Category'],data=df_install , orient='y' ,hue='Category')
ax.set_xlabel('No. of Installations in Billions')
ax.set_ylabel('')
ax.set_title("Most Popular Categories in Play Store", size = 20)
plt.show()
```

### 2. What are the Top 5 most installed Apps in Each popular Categories ??

```python
df.head(2)
```

```python
df_app = df_copy.groupby(['App','Category'])['Installs'].sum().reset_index()
df_a = df_app.sort_values('Installs',ascending=False)
```

```python
df_a.head(5)
```

```python
apps = ['GAME', 'COMMUNICATION', 'PRODUCTIVITY', 'SOCIAL' ]
plt.figure(figsize=(40,30))

for i,app in enumerate(apps):
    df2 = df_a[df_a.Category==app]
    df3 = df2.head(5)
    plt.subplot(4,2,i+1)
    sns.barplot(data=df3, x='Installs',y='App',hue='App')
    plt.xlabel('Installation in Millions')
    plt.ylabel('')
    plt.title(app,size = 20)
    
plt.tight_layout()
plt.subplots_adjust(hspace= .3)
plt.show()
```

## How many apps are there on Google Play Store which get 5 ratings??

```python
rating = df_copy.groupby(['Category','Installs', 'App'])['Rating'].sum().sort_values(ascending = False).reset_index()
```

```python
top_rated = rating[rating.Rating==5.0]
print("Number of 5 rated apps",top_rated.shape[0])
```
